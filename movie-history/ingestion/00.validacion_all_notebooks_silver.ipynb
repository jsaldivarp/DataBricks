```sql

--Tablas sin particionar
Select 'languages' tabla_name, file_date, count(1) from movie_silver.languages group by file_date
union all
Select 'genres' tabla_name, file_date, count(1) from movie_silver.genres group by file_date
union all
Select 'countries' tabla_name, file_date, count(1) from movie_silver.countries group by file_date
union all
Select 'languages_roles' tabla_name, file_date, count(1) from movie_silver.languages_roles group by file_date
```
