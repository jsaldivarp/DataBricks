```sql
CREATE OR REPLACE TEMP VIEW v_movies_2011 AS
select m.movie_id, m.title, m.budget, m.revenue, m.year_release_date, m.release_date
from movie_silver.movies m
where m.year_release_date = 2011
```

```sql
CREATE OR REPLACE TEMP VIEW v_productions_countries AS
select *
from movie_silver.productions_countries m
```

```sql
CREATE OR REPLACE TEMP VIEW v_countries AS
select c.country_id, c.country_name
from movie_silver.countries c
```

```sql
--INNER JOIN

select vm.movie_id, vm.title, vm.revenue, vm.year_release_date, vm.release_date, vc.country_name
from v_movies_2011 vm inner join v_productions_countries vpc
     on vm.movie_id = vpc.movie_id
     inner join v_countries vc
     on vc.country_id = vpc.country_id
```

```sql
--LEFT JOIN

select vm.movie_id, vm.title, vm.revenue, vm.year_release_date, vm.release_date, vc.country_name
from v_movies_2011 vm LEFT join v_productions_countries vpc
     on vm.movie_id = vpc.movie_id
     inner join v_countries vc
     on vc.country_id = vpc.country_id
```

```sql
--RIGHT JOIN

select vm.movie_id, vm.title, vm.revenue, vm.year_release_date, vm.release_date, vc.country_name
from v_movies_2011 vm right join v_productions_countries vpc
     on vm.movie_id = vpc.movie_id
     inner join v_countries vc
     on vc.country_id = vpc.country_id
```

```sql
--FULL JOIN

select vm.movie_id, vm.title, vm.revenue, vm.year_release_date, vm.release_date, vc.country_name
from v_movies_2011 vm FULL join v_productions_countries vpc
     on vm.movie_id = vpc.movie_id
     inner join v_countries vc
     on vc.country_id = vpc.country_id
```

```sql
--SEMI JOIN

select vm.movie_id, vm.title, vm.revenue, vm.year_release_date, vm.release_date--, vc.country_name
from v_movies_2011 vm INNER join v_productions_countries vpc
     on vm.movie_id = vpc.movie_id
     SEMI join v_countries vc
     on vc.country_id = vpc.country_id
```

```sql
--ANTI JOIN

select vm.movie_id, vm.title, vm.revenue, vm.year_release_date, vm.release_date--, vc.country_name
from v_movies_2011 vm INNER join v_productions_countries vpc
     on vm.movie_id = vpc.movie_id
     anti join v_countries vc
     on vc.country_id = vpc.country_id
```
