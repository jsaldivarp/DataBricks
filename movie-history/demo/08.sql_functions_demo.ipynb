#Scalar Funciones (Funciones Escalares)

```sql
use movie_silver;

--Funcion Concat
select m.*, concat(m.title, '-', m.release_date) as title_with_release_date
from movie_silver.movies m

```

```sql
--Funcion Concat
select p.*, split(p.name, ' '),
       get(split(p.name, ' '),0) as foremane,
       get(split(p.name, ' '), 2) as surname
from movie_silver.persons p
```

```sql
select current_timestamp() 
from movie_silver.movies m
```

```sql
select m.release_date , date_format(m.release_date,'dd/mm/yyyy')
from movie_silver.movies m
```

```sql
--agrega 1 dia
select m.release_date , date_add(m.release_date, 1)
from movie_silver.movies m
```

## Aggregate Functions 

```sql
select count(1)
from movie_silver.movies m
```

```sql
select max(m.release_date)
from movie_silver.movies m
```

```sql
select m.*
from movie_silver.movies m
where m.year_release_date = 2010
```

```sql
select m.year_release_date, count(1) total_count, sum(m.budget) sum_budget,
       max(m.budget) max_budget, min(m.budget) min_budget, round(avg(m.budget)) avg_budget
from movie_silver.movies m
group by m.year_release_date
having total_count > 220
order by 2;
```

#Windows Function (Funciones de Ventana)

```sql
select m.title, m.year_release_date, m.release_date,
       rank() over(partition by m.year_release_date order by m.release_date) as rank
from movie_silver.movies m
where m.year_release_date is not null 
order by release_date;
```
