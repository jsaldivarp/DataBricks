```python
movies_df = spark.read.parquet(f"{silver_folder_path}/movies")
```

```python
movies_df.filter("year_release_date = 2007 and movie_id = 35") .show(4)
```

```python
movies_df.filter("movie_id not in (35, 285)") .show(4)
```

```python
movies_df.filter("title like '%Four%'") .show(4)
```

```python
movies_df.filter("duration_time is null") .show(4)
```

```python
#cuando hay una comparacion como > entonces la expresion debe ir entre parentesis
from pyspark.sql.functions import col

movies_df.filter(
    col("duration_time").isNotNull() &
    (col("duration_time") > 100) &
    ~col("title").like("%Four%")
).show(4)
```

```python
#Operador "or" -> |

movies_df.where(
    col("duration_time").isNotNull() |
    col("title").like("%Four%")
).show(4)
```
