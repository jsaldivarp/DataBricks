```python

#DataFrame movies filtrado por el año 2007
movies_df = spark.read.parquet(f"{silver_folder_path}/movies")
movies_df = movies_df.filter(
                            (col("year_release_date") == 2007)
                            )
                            
#DataFrame Production Country
producion_country_df = spark.read.parquet(f"{silver_folder_path}/production_countries")

#DataFrame Countries
country_df = spark.read.parquet(f"{silver_folder_path}/countries")

```

```python
#Vemos los campos de cada tabla
movies_df.show(4)
producion_country_df.show(4)
country_df.show(4)
```

```python
#Generamos el Join por el campo movie id, como no se coloco nada, se asume que es un INNER
#Como no se indico nada entonces toma todos los campos de ambas tablas

movie_production_country_df = movies_df.join(producion_country_df, 
                                              movies_df.movie_id == producion_country_df.movie_id,
                                              "inner")\
                                        .select(movies_df.title, movies_df.budget, producion_country_df.country_id)

#movie_production_country_df.show(4)
```

```python
#Realiosamos el segundo join
movie_country_df = movie_production_country_df.join(country_df,
                                                    movie_production_country_df.country_id == country_df.country_id,
                                                    "inner")\
                                               .select(movie_production_country_df.title, movie_production_country_df.budget, country_df.country_name)
```

```python
display(movie_country_df)
```

# Outer Join

###Left Outer Join

```python
#Left Outer Join

movie_production_country_df = movies_df.join(producion_country_df, 
                                              movies_df.movie_id == producion_country_df.movie_id,
                                              "left")\
                                        .select(movies_df.title, movies_df.budget, producion_country_df.country_id)\
                                        #.filter(producion_country_df.movie_id.isNull())

display(movie_production_country_df)
```

```python
#Realiosamos el segundo join
movie_country_df = movie_production_country_df.join(country_df,
                                                    movie_production_country_df.country_id == country_df.country_id,
                                                    "left")\
                                               .select(movie_production_country_df.title, movie_production_country_df.budget, country_df.country_name)\
                                               .filter(country_df.country_id.isNull())

display(movie_country_df)
```

###Right Outer Join

```python
movie_production_country_df = movies_df.join(producion_country_df, 
                                              movies_df.movie_id == producion_country_df.movie_id,
                                              "right")\
                                        .select(movies_df.title, movies_df.budget, producion_country_df.country_id)\
                                        #.filter(producion_country_df.movie_id.isNull())

display(movie_production_country_df)
```

```python
movie_country_df = movie_production_country_df.join(country_df,
                                                    movie_production_country_df.country_id == country_df.country_id,
                                                    "right")\
                                               .select(movie_production_country_df.title, movie_production_country_df.budget, country_df.country_name)#\
                                               #.filter(country_df.country_id.isNull())

display(movie_country_df)
```

### Semi Join

```python
movie_production_country_df = movies_df.join(producion_country_df, 
                                              movies_df.movie_id == producion_country_df.movie_id,
                                              "left")\
                                        .select(movies_df.title, movies_df.budget, producion_country_df.country_id)#\
                                        #.filter(producion_country_df.movie_id.isNull())

display(movie_production_country_df)
```

```python
movie_country_df = movie_production_country_df.join(country_df,
                                                    movie_production_country_df.country_id == country_df.country_id,
                                                    "semi")\
                                               .select(movie_production_country_df.title, movie_production_country_df.budget)

display(movie_country_df)
```

### Anti Join

```python
movie_production_country_df = movies_df.join(producion_country_df, 
                                              movies_df.movie_id == producion_country_df.movie_id,
                                              "left")\
                                        .select(movies_df.title, movies_df.budget, producion_country_df.country_id)\
                                        .filter(producion_country_df.movie_id.isNull())

display(movie_production_country_df)
```

```python
movie_country_df = movie_production_country_df.join(country_df,
                                                    movie_production_country_df.country_id == country_df.country_id,
                                                    "anti")\
                                               .select(movie_production_country_df.title, movie_production_country_df.budget)#\
                                               #.filter(country_df.country_id.isNull())

display(movie_country_df)
```
