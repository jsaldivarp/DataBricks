###Local Temp View

```python
#Crear vista temporal en la base de datos
result_movie_genre_language_df = spark.read.parquet(f"{gold_folder_path}/results_movie_genre_language")

#vista temporal_local
result_movie_genre_language_df.createOrReplaceTempView("v_movie_genre_language")


```

```python
%sql
--Acceder a la vista desde celda SQL

 SELECT * 
 FROM v_movie_genre_language
 WHERE vote_average > 7.5
 
```

```python
#Acceder a la vista desde Phyton
#en phyton puedo guadar el resultado en un dataSet

result_movie_genre_language_2 = spark.sql("SELECT * FROM v_movie_genre_language WHERE vote_average > 7.5")
result_movie_genre_language_2.display()

#Tambien se puede poner parametros como p_vote_average
p_vote_average = 7.5
result_movie_genre_language_3 = spark.sql(f"SELECT * FROM v_movie_genre_language WHERE vote_average > {p_vote_average}")
result_movie_genre_language_3.display()
```

###Global Temp View

```python
#Crear vista temporal en la base de datos
result_movie_genre_language_df = spark.read.parquet(f"{gold_folder_path}/results_movie_genre_language")

#vista temporal_global
result_movie_genre_language_df.createOrReplaceGlobalTempView("gv_movie_genre_language")
```

```python
%sql
show tables in global_temp
```

```python
%sql
--Acceder a la vista desde celda SQL

 SELECT * 
 FROM global_temp.gv_movie_genre_language
 WHERE vote_average > 7.5
 
```

```python
#Acceder a la vista desde Phyton
#en phyton puedo guadar el resultado en un dataSet

result_movie_genre_language_2 = spark.sql("SELECT * FROM global_temp.gv_movie_genre_language WHERE vote_average > 7.5")
result_movie_genre_language_2.display()

#Tambien se puede poner parametros como p_vote_average
p_vote_average = 7.5
result_movie_genre_language_3 = spark.sql(f"SELECT * FROM global_temp.gv_movie_genre_language WHERE vote_average > {p_vote_average}")
result_movie_genre_language_3.display()
```
