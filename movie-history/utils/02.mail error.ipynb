```python
import smtplib
from email.mime.text import MIMEText

subject = 'PipeLine Error'
body = 'Error en el proceso de carga de pípeline, ya que no se encontro la carpeta'
sender = 'jsaldivarp@gmail.com'
my_password = 'qfnamnwfimnibrzx'
recipients = [sender]

def send_mail(subject, body, sender, recipients):
    msg = MIMEText(body)
    msg['subject'] = subject
    msg['From'] = sender
    msg['To'] = ', '.join(recipients)

    with smtplib.SMTP_SSL('smtp.gmail.com', 465) as smtp_server:
        smtp_server.login(sender, my_password)
        smtp_server.sendmail(sender, recipients, msg.as_string())
    print("Message sent!")

send_mail(subject, body, sender, recipients)
```
