#Crear tablas archivos JSON

```sql
--Archivo  country.json

use movie_bronze;

drop table if exists movie_bronze.country;
create table if not exists movie_bronze.country(
    countryId int,
    countryIsoCode string,
    countryName string
)
using json
options(path "abfss://bronze@lsdata01.dfs.core.windows.net/country.json")

```

```sql
select * from movie_bronze.country;
```

```sql
--Archivo JSON de 1 sola linea pero con una estructura compleja
--Archivo Person

use movie_bronze;
drop table if exists movie_bronze.persons;
create table if not exists movie_bronze.persons(
    personId int,
    personName struct < forename: string, surname: string>
)
using json
options(path "abfss://bronze@lsdata01.dfs.core.windows.net/person.json")
```

```sql
select * from movie_bronze.persons
```

```sql
--Archivo  movie_genre.json
use movie_bronze;
drop table if exists movie_bronze.movie_genres;
create table if not exists movie_bronze.movie_genres(
    movieId int,
    genreId int
)
using json
options(path "abfss://bronze@lsdata01.dfs.core.windows.net/movie_genre.json")
```

```sql
select * from movie_bronze.movie_genres
```

```sql
--Archivo JSON multilinea con estructura simple
use movie_bronze;
drop table if exists movie_bronze.movie_cast;
create table if not exists movie_bronze.movie_cast(
    movieId int,
    personId int,
    characterName string,
    genderId int,
    castOrder int
)
using json
options(path "abfss://bronze@lsdata01.dfs.core.windows.net/movie_cast.json", multiLine true)
```

```sql
select * from movie_bronze.movie_cast
```

```sql
--Archivo JSON multilinea con estructura simple
use movie_bronze;
drop table if exists movie_bronze.language_roles;
create table if not exists movie_bronze.language_roles(
    roleId int,
    languageRole string
)
using json
options(path "abfss://bronze@lsdata01.dfs.core.windows.net/language_role.json", multiLine true)
```

```sql

select * from movie_bronze.language_roles;
```
