```sql
use movie_gold;
```

```sql
--Presupuesto ingreso por productora
select rm.company_name,
       count(rm.company_name) as total_movie,
       sum(rm.budget)  as total_budget,
       cast(avg(rm.budget) as decimal(18,2)) as avg_budget,
       sum(rm.revenue) as total_revenue,
       cast(avg(rm.revenue) as decimal(18,2)) as avg_revenue 
from movie_gold.results_movie rm
where rm.year_release_date = 2010 
group by rm.company_name
order by total_budget desc
limit 10;
```
